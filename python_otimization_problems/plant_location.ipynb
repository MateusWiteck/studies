# Plant location models

### **Context:**
A logistics company is considering opening plants at several potential locations to meet customer demand at various cities. Due to uncertainty in demand and transportation costs, the company wants to plan for different scenarios that reflect various economic conditions or fluctuations in customer needs.

#### Sets:
- $\mathcal{P}$: Set of potential plant locations.
- $\mathcal{D}$: Set of demand nodes (cities where customers are located).
- $\mathcal{S}$: Set of scenarios (different demand and transportation cost scenarios).

#### Parameters:
- $f_i$: Fixed cost of opening plant $i \in \mathcal{P}$.
- $R_i$: Capacity of plant $i \in \mathcal{P}$.
- $d_j^s$: Demand at node $j \in \mathcal{D}$ in scenario $s \in \mathcal{S}$.
- $c_{ij}^s$: Unit transportation cost between plant $i$ and node $j$ in scenario $s \in \mathcal{S}$.
- $p_s$: Probability of scenario $s \in \mathcal{S}$ occurring.

#### Decision Variables:
- $y_i \in \{0, 1\}$: Binary variable indicating whether plant $i \in \mathcal{P}$ is opened.
- $x_{ij}^s \geq 0$: Amount of product transported from plant $i$ to demand node $j$ in scenario $s \in \mathcal{S}$.

#### Objective:
The goal is to minimize the total expected cost, which includes the fixed cost of opening plants and the expected transportation cost over all scenarios.

### **Stochastic Plant Location Model:**
\[
\text{min} \quad \sum_{i \in \mathcal{P}} f_i y_i + \sum_{s \in \mathcal{S}} p_s \left( \sum_{i \in \mathcal{P}} \sum_{j \in \mathcal{D}} c_{ij}^s x_{ij}^s \right)
\]
subject to:
1. **Demand satisfaction constraint for each scenario**:
\[
\sum_{i \in \mathcal{P}} x_{ij}^s \geq d_j^s, \quad \forall j \in \mathcal{D}, \forall s \in \mathcal{S}
\]
2. **Capacity constraint**:
\[
\sum_{j \in \mathcal{D}} x_{ij}^s \leq R_i y_i, \quad \forall i \in \mathcal{P}, \forall s \in \mathcal{S}
\]
3. **Non-negativity and binary constraints**:
\[
x_{ij}^s \geq 0, \quad \forall i \in \mathcal{P}, \forall j \in \mathcal{D}, \forall s \in \mathcal{S}
\]
\[
y_i \in \{0, 1\}, \quad \forall i \in \mathcal{P}
\]

### **Data Example**:

#### Set of Plant Locations ($\mathcal{P}$):
- Plant 1
- Plant 2
- Plant 3

#### Set of Demand Nodes ($\mathcal{D}$):
- City A
- City B
- City C

#### Set of Scenarios ($\mathcal{S}$):
- Scenario 1 (high demand, low transportation cost)
- Scenario 2 (low demand, high transportation cost)
- Scenario 3 (medium demand, medium transportation cost)

#### Fixed Plant Opening Costs ($f_i$):
- $f_1 = 1000$
- $f_2 = 1200$
- $f_3 = 900$

#### Plant Capacities ($R_i$):
- $R_1 = 500$
- $R_2 = 600$
- $R_3 = 400$

#### Demand Data ($d_j^s$):
- Scenario 1: $d_A^1 = 200$, $d_B^1 = 150$, $d_C^1 = 100$
- Scenario 2: $d_A^2 = 100$, $d_B^2 = 80$, $d_C^2 = 60$
- Scenario 3: $d_A^3 = 150$, $d_B^3 = 120$, $d_C^3 = 90$

#### Transportation Costs ($c_{ij}^s$):
- Scenario 1: Low costs
  - $c_{11}^1 = 2$, $c_{12}^1 = 3$, $c_{13}^1 = 5$
  - $c_{21}^1 = 4$, $c_{22}^1 = 2$, $c_{23}^1 = 3$
  - $c_{31}^1 = 3$, $c_{32}^1 = 5$, $c_{33}^1 = 4$

- Scenario 2: High costs
  - $c_{11}^2 = 6$, $c_{12}^2 = 7$, $c_{13}^2 = 9$
  - $c_{21}^2 = 8$, $c_{22}^2 = 5$, $c_{23}^2 = 6$
  - $c_{31}^2 = 7$, $c_{32}^2 = 9$, $c_{33}^2 = 8$

- Scenario 3: Medium costs
  - $c_{11}^3 = 4$, $c_{12}^3 = 5$, $c_{13}^3 = 7$
  - $c_{21}^3 = 6$, $c_{22}^3 = 4$, $c_{23}^3 = 5$
  - $c_{31}^3 = 5$, $c_{32}^3 = 7$, $c_{33}^3 = 6$

#### Scenario Probabilities ($p_s$):
- Scenario 1: $p_1 = 0.5$
- Scenario 2: $p_2 = 0.3$
- Scenario 3: $p_3 = 0.2$


This setup can be solved using stochastic optimization techniques like scenario-based linear programming or decomposition methods such as Benders' decomposition.

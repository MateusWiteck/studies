```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np

cost = 1
selling_price = 10 
pi = [0.4, 0.3, 0.21, 0.09]
vals = [0,1,2,3]
n_scenarios = len(vals)
scenarios = range(n_scenarios)

# Model
m = gp.Model("newsvendor")

n_newspaper = m.addVar(vtype=GRB.INTEGER, lb=0, name="X")
y = m.addMVar(shape=n_scenarios, vtype=GRB.INTEGER, lb=0, name="Y")

# Objective: Maximize expected profit
## In each scenario we have the following profit:
profit = [(selling_price * y[s] - cost * n_newspaper) for s in scenarios]
## Now we maximize the expected profit for each scenario ponderating each probability:
m.setObjective(
    gp.quicksum(pi[s] * profit[s] for s in scenarios),
    GRB.MAXIMIZE
)

# Constraints
for s in scenarios:
    m.addConstr(y[s] <= n_newspaper) # Can't sell more newspaper than the number of buyed nespaper
    m.addConstr(y[s] <= vals[s])     # Can't sell more newspaper than the demmand in the referred scenario

# Save model
# m.write("newsvendor.lp")

# Solve
m.optimize()

# Output results to analyse each scenario
if m.status == GRB.OPTIMAL:
    print(f"Optimal number of newspapers to buy: {n_newspaper.X}")
    for s in scenarios:
        print(f"Scenario {s}: Sold {y[s].X} newspapers")
```

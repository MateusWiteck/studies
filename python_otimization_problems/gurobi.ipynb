# Introduction to Gurobi in Python

The code below test if your license is working. If it's not working you need to install the gurobi software  and activate with your license. 

```python
import gurobipy as gp
from gurobipy import GRB

def test_gurobi_license():
    try:
        # Create a simple optimization model to test the license
        model = gp.Model("test_license")

        # Add a decision variable
        x = model.addVar(name="x", lb=0, ub=1, vtype=GRB.CONTINUOUS)

        # Set the objective function (maximize x)
        model.setObjective(x, GRB.MAXIMIZE)

        # Add a constraint (x <= 0.5)
        model.addConstr(x <= 0.5, name="constraint")

        # Optimize the model
        model.optimize()

        # Check the optimization result
        if model.status == GRB.OPTIMAL:
            print("License test passed! Optimization was successful.")
        else:
            print(f"License test failed. Model status: {model.status}")

    except gp.GurobiError as e:
        print(f"Gurobi error: {e}")

    except Exception as e:
        print(f"General error: {e}")

test_gurobi_license()
```

## What problems can you solve with gurobi?

### Linear Programming (LP)

**Linear Programming (LP)** is one of the most fundamental types of optimization problems, where the objective is to maximize or minimize a linear function, subject to linear equality or inequality constraints. The variables involved in LP are continuous and must satisfy the constraints, which form a convex feasible region. The solution to an LP problem lies at one of the vertices of this region. LP is widely used in areas like resource allocation, production planning, and logistics. One of the key advantages of LP is its computational efficiency, especially for large-scale problems, as specialized algorithms like the Simplex method or Interior Point methods can solve them quickly.

### Mixed-Integer Programming (MIP)

**Mixed-Integer Programming (MIP)** extends linear programming by allowing some or all of the decision variables to be integers. This introduces a layer of complexity because the solution space becomes discrete, and standard LP methods do not directly apply. MIP is particularly useful in real-world situations where decisions are inherently binary or require whole numbers, such as deciding whether to build a factory (binary decision) or determining the number of units to produce (integer decision). Solving MIP problems often involves combinatorial optimization techniques like branch-and-bound or branch-and-cut. MIP is frequently used in supply chain management, transportation, and scheduling problems where decisions must be made in a structured, discrete manner.

### Quadratic Programming (QP)

**Quadratic Programming (QP)** is another extension of linear programming where the objective function includes quadratic terms, allowing the optimization of a function that involves variables raised to the power of two (such as \( x_1^2 \)). The constraints in QP remain linear, but the quadratic objective introduces non-linearity into the model. QP is particularly useful in scenarios where there is a need to model risk or interaction effects between variables, such as in financial portfolio optimization, where one might seek to minimize the risk (measured by variance) while maintaining a certain level of return. QP problems can still be convex if the quadratic part of the objective function is convex, making them solvable by efficient algorithms.

### Stochastic Programming

**Stochastic Programming** is an optimization framework that deals with uncertainty in the problem data, which is typically represented by random variables. In stochastic programming, instead of having fixed values for all parameters, some variables have probability distributions, and the objective is to make decisions that are optimal on average across several possible future scenarios. Stochastic programming is particularly relevant in fields like finance, energy planning, and supply chain management, where future events are uncertain (e.g., demand fluctuations, stock prices, or weather conditions). Common approaches include two-stage stochastic programming, where decisions are made in two phases: one before the uncertainty is revealed and one after, allowing for adaptation. By considering a wide range of scenarios, stochastic programming models help in creating robust, resilient decision-making strategies that can handle variability.
